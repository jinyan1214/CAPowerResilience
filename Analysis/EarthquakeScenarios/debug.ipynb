```python
import os
import pandas as pd
import numpy as np
from tqdm import tqdm
```

```python
pypsa_data_dir = "/resnick/groups/enceladus/jyzhao/CAPowerResilience/Data/eq_handoff_v5/handoff"
expansion_scenario_list = ["conservative", "bau", "aggressive"]
agg_type = "disagg"
eq_scenario_id = "bssc2014haywardhnhsshaw09mod_m7p32_se"
eq_scenario_dir = f"/resnick/groups/enceladus/jyzhao/CAPowerResilience/Data/BSSC2014_scenarios/{eq_scenario_id}"
pga_samples_all = pd.read_csv(os.path.join(eq_scenario_dir, "sites_gm_samples.csv"))
```

```python
scenario_i = 0
expansion_scenario = expansion_scenario_list[scenario_i]
```

```python
assets_df = pd.read_csv(os.path.join(pypsa_data_dir,expansion_scenario,"disagg",
 "assets_all_aug.csv"))


buses_df = pd.read_csv(os.path.join(pypsa_data_dir,expansion_scenario,"disagg",
 "buses_all_aug.csv"))
print(buses_df.shape)

assets_df['asset_id'] = range(len(assets_df)) 
assets_df['asset_id'] += len(buses_df)
```

```python
pga_samples_df = pd.read_csv(
    f"/resnick/groups/enceladus/jyzhao/CAPowerResilience/Data/BSSC2014_scenarios/{eq_scenario_id}/{expansion_scenario}/pga_samples.csv",
    index_col=0)
pga_samples_df.head()
```

```python
## Verify PGA Samples DataFrame
for i in tqdm(range(assets_df.shape[0])):
    site_id = assets_df.loc[i, 'site_id']
    asset_id = assets_df.loc[i, 'asset_id']
    pga_samples = pga_samples_df[str(asset_id)].values[1:].astype(float)

    pga_samples_true =pga_samples_all[str(site_id)].values

    if not np.allclose(pga_samples, pga_samples_true):
        print(f"Mismatch for asset_id {asset_id} and site_id {site_id}")
```

```python
demand_df = pd.read_csv(
    f"/resnick/groups/enceladus/jyzhao/CAPowerResilience/Data/BSSC2014_scenarios/{eq_scenario_id}/{expansion_scenario}/demand.csv",
    index_col=0)
demand_df.head()
```
