```python
import os
import pandas as pd
```

```python
pypsa_data_dir = "/resnick/groups/enceladus/jyzhao/CAPowerResilience/Data/eq_handoff_v5/handoff"
expansion_scenario_list = ["conservative", "bau", "aggressive"]
agg_type = "disagg"
```

```python
buses_list = []
# Buses / substations
for expansion_scenario in expansion_scenario_list:
    buses_all_fp = os.path.join(
        pypsa_data_dir,
        expansion_scenario,
        agg_type,
        "buses_all.csv"
    )
    buses_df = pd.read_csv(buses_all_fp)
    buses_list.append(buses_df)
```

```python
# Check if the buses in each scenario are the same
for i in range(1, len(buses_list)):
    common_buses = set(buses_list[0]['bus']).intersection(set(buses_list[i]['bus']))
    print(f"Common buses between scenario 0 and scenario {i}: {len(common_buses)}")
    if len(common_buses) != len(buses_list[0]['bus']):
        print(f"Scenario {i} has different buses than scenario 0")
        different_buses = set(buses_list[i]['bus']).difference(set(buses_list[0]['bus']))
        print(f"Buses in scenario {i} but not in scenario 0: {different_buses}")
        missing_buses = set(buses_list[0]['bus']).difference(set(buses_list[i]['bus']))
        print(f"Buses in scenario 0 but not in scenario {i}: {missing_buses}")
```

```python
assets_list = []
# Assets
for expansion_scenario in expansion_scenario_list:
    assets_all_fp = os.path.join(
        pypsa_data_dir,
        expansion_scenario,
        agg_type,
        "assets_all.csv"
    )
    assets_df = pd.read_csv(assets_all_fp)
    assets_list.append(assets_df)
```

```python
# Check if bau and aggressive contains all conservative
for i in range(1, len(assets_list)):
    common_assets = set(assets_list[0]['asset_key']).intersection(set(assets_list[i]['asset_key']))
    print(f"Common assets between conservative and {expansion_scenario_list[i]}: {len(common_assets)}")
    if len(common_assets) != len(assets_list[0]['asset_key']):
        print(f"Scenario {expansion_scenario_list[i]} has different assets than conservative")
        different_assets = set(assets_list[i]['asset_key']).difference(set(assets_list[0]['asset_key']))
        print(f"Number of assets in {expansion_scenario_list[i]} but not in conservative: {len(different_assets)}")
        missing_assets = set(assets_list[0]['asset_key']).difference(set(assets_list[i]['asset_key']))
        print(f"Number of assets in conservative but not in {expansion_scenario_list[i]}: {len(missing_assets)}")
```

```python
# Check if aggressive scenario contains all assets from the bau scenario
for i in range(2, len(assets_list)):
    common_assets = set(assets_list[1]['asset_key']).intersection(set(assets_list[i]['asset_key']))
    print(f"Common assets between bau and {expansion_scenario_list[i]}: {len(common_assets)}")
    if len(common_assets) != len(assets_list[1]['asset_key']):
        print(f"Scenario {expansion_scenario_list[i]} has different assets than bau")
        different_assets = set(assets_list[i]['asset_key']).difference(set(assets_list[1]['asset_key']))
        print(f"Number of assets in {expansion_scenario_list[i]} but not in bau: {len(different_assets)}")
        missing_assets = set(assets_list[1]['asset_key']).difference(set(assets_list[i]['asset_key']))
        print(f"Number of assets in bau but not in {expansion_scenario_list[i]}: {len(missing_assets)}")
```

```python
# Find the common assets in all three asset expansion scenarios
common_assets = assets_list[0]
for i in range(1, len(assets_list)):
    common_assets = common_assets[common_assets['asset_key'].isin(assets_list[i]['asset_key'])]
print(len(common_assets))
```

```python
# Find the common assets in all three asset expansion scenarios
common_buses = buses_list[0]
for i in range(1, len(buses_list)):
    common_buses = pd.merge(common_buses, buses_list[i], on='bus', how='inner')
print(len(common_buses))
```
